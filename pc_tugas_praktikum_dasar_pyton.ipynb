```python

# TUGAS 2: MENGOLAH STRING

# Menyimpan nama depan dan nama belakang
nama_depan = "Firman"
nama_belakang = "Abidin"

# Menggabungkan nama depan dan nama belakang
nama_lengkap = nama_depan + " " + nama_belakang

# Menyimpan sebuah kalimat
kalimat = "saya sedang belajar bahasa python dasar"

# Menampilkan nama lengkap
print("=== DATA NAMA ===")
print("Nama depan:", nama_depan)
print("Nama belakang:", nama_belakang)
print("Nama lengkap:", nama_lengkap)

# Mengubah format huruf pada kalimat
print("\n=== OPERASI STRING ===")
print("Kalimat asli:", kalimat)
print("Uppercase:", kalimat.upper())
print("Lowercase:", kalimat.lower())
print("Capitalize:", kalimat.capitalize())
print("Title:", kalimat.title())

# Menghitung jumlah kemunculan huruf a
jumlah_huruf_a = kalimat.count("a")
print("Jumlah huruf 'a':", jumlah_huruf_a)
```

```python

# TUGAS 3: OPERASI MATEMATIKA

# Meminta pengguna memasukkan dua angka
angka_pertama = float(input("Masukkan angka pertama: "))
angka_kedua = float(input("Masukkan angka kedua: "))

# Melakukan operasi matematika
penjumlahan = angka_pertama + angka_kedua
pengurangan = angka_pertama - angka_kedua
perkalian = angka_pertama * angka_kedua
pangkat = angka_pertama ** angka_kedua

# Menampilkan hasil kalkulator
print("\n=== HASIL KALKULATOR ===")
print("Penjumlahan:", penjumlahan)
print("Pengurangan:", pengurangan)
print("Perkalian:", perkalian)

# Memeriksa agar tidak terjadi pembagian dengan nol
if angka_kedua != 0:
    pembagian = angka_pertama / angka_kedua
    print("Pembagian:", pembagian)
else:
    print("Pembagian: tidak dapat dilakukan karena pembagi nol")

print("Pangkat:", pangkat)

# Meminta input panjang dan lebar persegi panjang
panjang = float(input("\nMasukkan panjang persegi panjang: "))
lebar = float(input("Masukkan lebar persegi panjang: "))

# Menghitung luas persegi panjang
luas = panjang * lebar

# Menampilkan hasil luas
print("\n=== LUAS PERSEGI PANJANG ===")
print("Panjang:", panjang)
print("Lebar:", lebar)
print("Luas persegi panjang:", luas)
```

```python
# ==========================================
# TUGAS 4 — Perulangan for dan while
# ==========================================

# 1. Membuat daftar minimal lima nama mahasiswa
daftar_mahasiswa = ['Firman', 'Ahmad', 'Siti', 'Dewi', 'Budi']

print("=== 1. Tampilkan Nama Menggunakan 'for' ===")
for nama in daftar_mahasiswa:
    print(f"- {nama}")

print()

print("=== 2. Tampilkan Angka 1 sampai 10 (range()) ===")
for angka in range(1, 11):
    print(angka, end=" ")  # Tampil menyamping
print("\n")

print("=== 3. Tampilkan Nama Menggunakan 'while' & len() ===")
i = 0
while i < len(daftar_mahasiswa):
    print(f"Mahasiswa ke-{i + 1}: {daftar_mahasiswa[i]}")
    i += 1

print()

print("=== 4. Bilangan Genap 2 sampai 20 ('while') ===")
bilangan_genap = 2
while bilangan_genap <= 20:
    print(bilangan_genap, end=" ")
    bilangan_genap += 2

print("\n")
```

```python

# TUGAS 5: FUNGSI DAN WAKTU

# Mengimpor modul datetime
import datetime

# Membuat fungsi untuk mencatat waktu aktivitas
def catat_waktu():
    # Menampilkan pesan aktivitas selesai
    print("Aktivitas selesai")

    # Menampilkan waktu saat fungsi dijalankan
    waktu_sekarang = datetime.datetime.now()
    print("Waktu pelaksanaan:", waktu_sekarang)
    print("-" * 40)


# Mencatat waktu sebelum pengerjaan soal
print("=== MULAI BELAJAR ===")
catat_waktu()

# Melakukan perulangan untuk mengerjakan lima soal
print("=== PROSES PENGERJAAN SOAL ===")

for nomor_soal in range(1, 6):
    print("Sedang mengerjakan soal nomor", nomor_soal)

print("Semua 5 soal telah selesai dikerjakan.")
print("-" * 40)

# Mencatat waktu setelah pengerjaan soal
print("=== SELESAI BELAJAR ===")
catat_waktu()
```

```python
from datetime import datetime

# 8. Fungsi untuk menampilkan total dan rata-rata nilai
def tampilkan_hasil(total, rata_rata):
    print("--------------------------------")
    print(f"Total Nilai : {total}")
    print(f"Rata-rata   : {rata_rata}")

# 9. Fungsi untuk menampilkan waktu saat program menyelesaikan proses
def catat_waktu():
    # Mengambil waktu saat ini
    waktu_sekarang = datetime.now()
    # Format waktu: Hari, Tanggal Bulan Tahun - Jam:Menit:Detik
    format_waktu = waktu_sekarang.strftime("%d-%m-%Y %H:%M:%S")
    print(f"Waktu       : {format_waktu}")

# 1. Tampilkan judul program menggunakan print() dan garis pemisah
print("================================")
print("      REKAP NILAI MAHASISWA     ")
print("================================")

# 2. Minta input nama dan NIM mahasiswa
nama_input = input("Masukkan Nama Mahasiswa : ")
nim_input = input("Masukkan NIM Mahasiswa  : ")

# 5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan upper()
nama_kapital = nama_input.upper()
nim_bersih = nim_input.strip()

print("--------------------------------")

# 3. Minta pengguna memasukkan nilai tiga mata kuliah
# Menggunakan konversi float/int agar bisa dihitung
nilai_algo = float(input("Nilai Algoritma     : "))
nilai_db = float(input("Nilai Basis Data    : "))
nilai_struktur = float(input("Nilai Struktur Data : "))

# Menyimpan nilai ke dalam list untuk keperluan perulangan
daftar_nilai = [nilai_algo, nilai_db, nilai_struktur]
daftar_matkul = ["Nilai Algoritma", "Nilai Basis Data", "Nilai Struktur Data"]

print("--------------------------------")
print("HASIL REKAP:")
print(f"Nama       : {nama_kapital}")
print(f"NIM        : {nim_bersih}")
print("--------------------------------")

# 7. Gunakan while untuk menampilkan nomor urut mata kuliah dan nilainya bersamaan
i = 0
while i < len(daftar_matkul):
    # 6. Gunakan for (atau diintegrasikan dengan data) untuk menampilkan nilai
    # Di sini kita tampilkan sesuai nomor urut perulangan while
    print(f"{daftar_matkul[i]}: {daftar_nilai[i]}")
    i += 1

# 4. Hitung total dan rata-rata nilai menggunakan operasi matematika
total_nilai = sum(daftar_nilai)
rata_rata_nilai = total_nilai / len(daftar_nilai)

# Memanggil fungsi untuk menampilkan hasil total dan rata-rata (No. 8)
tampilkan_hasil(total_nilai, rata_rata_nilai)

print("--------------------------------")
print("Status      : Rekap selesai")

# Memanggil fungsi untuk menampilkan waktu (No. 9)
catat_waktu()
print("================================")
```
