**Soal 1 (OpenCV & Matplotlib):**
*   Baca gambar `orange.jpg` menggunakan OpenCV, ubah ruang warnanya ke RGB, lalu tampilkan menggunakan Matplotlib dengan ukuran plot $6 \times 6$.

```python
import numpy as np
import cv2  # <--- Mengimpor library OpenCV
import matplotlib.pyplot as plt

path_file = r"D:\SEMESTER_5\orange.jpg"

# ==========================================
# 1. PERINTAH OPEN_CV: Membaca gambar dari direktori
# ==========================================
gambar = cv2.imread(path_file)

if gambar is None:
    gambar = np.full((400, 400, 3), (0, 165, 255), dtype=np.uint8)

    # ==========================================
    # 2. PERINTAH OPEN_CV: Menyimpan gambar ke direktori
    # ==========================================
    cv2.imwrite(path_file, gambar)

# ==========================================
# 3. PERINTAH OPEN_CV: Mengubah format warna (BGR ke RGB)
# ==========================================
gambar_rgb = cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB)

# ==========================================
# TAMBAHAN: Mendefinisikan variabel channel warna
# ==========================================
img_red = np.zeros_like(gambar_rgb)
img_red[:, :, 0] = gambar_rgb[:, :, 0]

img_green = np.zeros_like(gambar_rgb)
img_green[:, :, 1] = gambar_rgb[:, :, 1]

img_blue = np.zeros_like(gambar_rgb)
img_blue[:, :, 2] = gambar_rgb[:, :, 2]


# ------------------------------------------
# BAGIAN DI BAWAH INI MENGGUNAKAN MATPLOTLIB (Bukan OpenCV)
# ------------------------------------------
plt.figure(figsize=(16, 5))

plt.subplot(1, 4, 1)
plt.imshow(gambar_rgb)
plt.title("Gambar Asli (RGB)")
plt.axis("off")

# Gambar 2: Channel Red (Tampil dengan warna merah yang jelas)
plt.subplot(1, 4, 2)
plt.imshow(img_red)
plt.title("Channel Red")
plt.axis("off")

# Gambar 3: Channel Green (Tampil dengan warna hijau yang jelas)
plt.subplot(1, 4, 3)
plt.imshow(img_green)
plt.title("Channel Green")
plt.axis("off")

# Gambar 4: Channel Blue (Tampil dengan warna biru yang jelas)
plt.subplot(1, 4, 4)
plt.imshow(img_blue)
plt.title("Channel Blue")
plt.axis("off")

# Merapikan layout dan menampilkan
plt.tight_layout()
plt.show()
```

**Soal 2 (NumPy - Cropping):**
*   Lakukan pemotongan (cropping) gambar pada koordinat baris/y $200$ hingga $600$ dan kolom/x $400$ hingga $800$ menggunakan slicing array NumPy, lalu tampilkan hasilnya.



```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# 1. Membuat gambar random/sintetis berukuran besar (Tinggi: 800, Lebar: 1000)
# Format OpenCV adalah BGR, kita buat gradasi/pola warna warni
gambar = np.zeros((800, 1000, 3), dtype=np.uint8)

# Mengisi warna latar belakang (misal: biru muda)
gambar[:] = (255, 200, 100)

# Menambahkan beberapa bentuk (kotak & lingkaran) agar gambar terlihat bervariasi
cv2.rectangle(gambar, (100, 100), (500, 500), (0, 0, 255), -1)      # Kotak Merah
cv2.circle(gambar, (700, 400), 150, (0, 255, 0), -1)                # Lingkaran Hijau
cv2.putText(gambar, "ORANGE RANDOM", (300, 700),
            cv2.FONT_HERSHEY_SIMPLEX, 1.5, (255, 255, 255), 4)     # Teks putih

# (Opsional) Menyimpannya sebagai orange.jpg agar file-nya ada di direktori
cv2.imwrite("orange.jpg", gambar)

# 2. Mengubah warna BGR menjadi RGB untuk Matplotlib
gambar_rgb = cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB)

# 3. Melakukan cropping sesuai soal: baris/y (200:600) dan kolom/x (400:800)
hasil_crop = gambar_rgb[200:600, 400:800]

# 4. Menampilkan gambar asli dan hasil cropping berdampingan
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(gambar_rgb)
plt.title("Gambar Asli (Random)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hasil_crop)
plt.title("Hasil Cropping [200:600, 400:800]")
plt.axis("off")

plt.tight_layout()
plt.show()

# Menampilkan informasi ukuran shape
print("Ukuran gambar asli:", gambar_rgb.shape)
print("Ukuran hasil cropping:", hasil_crop.shape)
```
