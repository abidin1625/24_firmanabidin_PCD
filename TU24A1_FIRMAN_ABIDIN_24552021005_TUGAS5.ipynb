## Tugas Praktikum 1: Eksperimen Nilai Ambang Batas (Thresholding)

**Tujuan:**
Memahami pengaruh perubahan nilai ambang batas (*threshold*) terhadap hasil citra biner.

**Instruksi Tugas:**
1. Gunakan citra ***selain `kucing.jpg`***.
2. Lakukan konversi citra tersebut menjadi Grayscale.
3. Buatlah **tiga variasi** nilai ambang batas (*threshold*), yaitu:
   - $T_1 = 80$
   - $T_2 = 128$ (standar)
   - $T_3 = 200$
4. Lakukan proses thresholding untuk masing-masing nilai ambang batas tersebut.
5. Tampilkan hasil ketiga citra biner tersebut secara berdampingan dalam satu baris menggunakan `matplotlib` (gunakan fungsi `plt.subplot`).
6. Tulis analisis singkat di bawah hasil visualisasi mengenai perbedaan yang terlihat pada ketiga gambar hasil thresholding tersebut.

**CONTOH HASIL**
image.png

```python
# @title Code jawaban Tugas Praktikum 1

#tulis disini

import cv2
import matplotlib.pyplot as plt

# 1. Load citra selain kucing (ganti 'bunga.jpg' dengan nama file citra Anda)
image_path = 'panda.jpg'  # Pastikan file gambar ada di direktori yang sama
img_color = cv2.imread(image_path)

# Pastikan gambar berhasil dibaca
if img_color is None:
    raise ValueError("File gambar tidak ditemukan. Periksa kembali path atau nama file.")

# 2. Konversi citra menjadi Grayscale
img_gray = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)

# 3 & 4. Tentukan 3 variasi nilai threshold dan lakukan proses thresholding
# Menggunakan cv2.THRESH_BINARY: piksel > threshold menjadi 255 (putih), sisanya 0 (hitam)
ret, thresh1 = cv2.threshold(img_gray, 80, 255, cv2.THRESH_BINARY)   # T1 = 80
ret, thresh2 = cv2.threshold(img_gray, 128, 255, cv2.THRESH_BINARY)  # T2 = 128 (standar)
ret, thresh3 = cv2.threshold(img_gray, 200, 255, cv2.THRESH_BINARY)  # T3 = 200

# 5. Tampilkan hasil ketiga citra biner secara berdampingan menggunakan plt.subplot
plt.figure(figsize=(15, 5))

# Subplot 1: Threshold = 80
plt.subplot(1, 3, 1)
plt.imshow(thresh1, cmap='gray')
plt.title('Threshold (T1) = 80')
plt.axis('off')

# Subplot 2: Threshold = 128
plt.subplot(1, 3, 2)
plt.imshow(thresh2, cmap='gray')
plt.title('Threshold (T2) = 128 (Standar)')
plt.axis('off')

# Subplot 3: Threshold = 200
plt.subplot(1, 3, 3)
plt.imshow(thresh3, cmap='gray')
plt.title('Threshold (T3) = 200')
plt.axis('off')

# Tampilkan plot secara keseluruhan
plt.tight_layout()
plt.show()
```

### Soal Praktikum 2: Menampilkan Channel Warna Tunggal (Color Channel Isolation)

**Instruksi:**
1. Buatlah program untuk mengisolasi kanal warna **Biru (Blue)** saja dari citra yang sudah diinputkan.
2. Ingat bahwa untuk menampilkan kanal biru murni (bukan grayscale), Anda harus menyetel nilai kanal Merah (R) dan Hijau (G) menjadi `0` pada array citra asli.
3. Tampilkan citra asli berdampingan dengan citra hasil isolasi kanal biru tersebut.

**CONTOH HASIL**
image.png

```python
# @title Code jawaban Tugas Praktikum 2

#tulis disini

import cv2
import matplotlib.pyplot as plt
import numpy as np

# 1. Load citra
image_path = 'panda.jpg'
img_color = cv2.imread(image_path)

# gambar berhasil dibaca
if img_color is None:
    raise ValueError("File gambar tidak ditemukan. Periksa kembali path atau nama file.")

# Opencv baca citra format BGR (Blue, Green, Red)
b, g, r = cv2.split(img_color)

# 2. Mengisolasi kanal Biru (Blue) :
# Buat citra kosong dengan ukuran dan data sama atau citra asli (berisi nilai 0)
zeros = np.zeros_like(b)

# menggabungkan kembali dengan menyetel kanal Red dan Green menjadi 0 (kosong)
# Format penggabungan OpenCV: [Blue, Green, Red]
img_blue_isolated = cv2.merge([b, zeros, zeros])

# 3. Tampilkan citra asli berdampingan dengan citra hasil isolasi kanal biru
# Konversi BGR ke RGB
img_rgb = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
img_blue_rgb = cv2.cvtColor(img_blue_isolated, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

# 1: Citra Asli
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Citra Asli')
plt.axis('off')

# 2: Citra Hasil Isolasi Kanal Biru
plt.subplot(1, 2, 2)
plt.imshow(img_blue_rgb)
plt.title('Isolasi Kanal Biru (Blue Only)')
plt.axis('off')

# plot secara keseluruhan
plt.tight_layout()
plt.show()
```
